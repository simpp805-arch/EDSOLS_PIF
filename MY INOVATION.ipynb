```python
import time
import speech_recognition as sr
import Arm_Lib

# Initialize the 6-DOF arm
Arm = Arm_Lib.Arm_Device()
time.sleep(0.5)

# Current joint positions: [J1, J2, J3, J4, J5, J6]
# Starts at a balanced, neutral center position
current_joints = [90, 90, 90, 90, 90, 30]

# Motion step sizes in degrees
STEP = 15

# Joint angle safety limits (DOFBOT limits: 0 to 180 degrees)
LIMITS = {
    "min": [0, 0, 0, 0, 0, 0],
    "max": [180, 180, 180, 180, 180, 180]
}

def clamp(val, min_val, max_val):
    return max(min_val, min(val, max_val))

def apply_pose(duration=500):
    """Enforces joint limits and sends command to the servos."""
    for i in range(6):
        current_joints[i] = clamp(current_joints[i], LIMITS["min"][i], LIMITS["max"][i])
    Arm.Arm_serial_servo_write6_array(current_joints, duration)
    time.sleep(duration / 1000.0)

# Apply initial neutral pose
apply_pose(1000)

def execute_voice_command(cmd):
    """Parses spoken text and translates it to arm directional shifts."""
    global current_joints
    cmd = cmd.lower()
    print(f"Executing: '{cmd}'")

    # --- Cardinal Directions ---
    if "left" in cmd and "diagonal" not in cmd:
        # Rotate base counter-clockwise
        current_joints[0] += STEP
    elif "right" in cmd and "diagonal" not in cmd:
        # Rotate base clockwise
        current_joints[0] -= STEP
    elif "up" in cmd and "diagonal" not in cmd:
        # Lift shoulder & elbow
        current_joints[1] += STEP
        current_joints[2] -= STEP
    elif "down" in cmd and "diagonal" not in cmd:
        # Lower shoulder & elbow
        current_joints[1] -= STEP
        current_joints[2] += STEP
    elif "forward" in cmd or "front" in cmd:
        # Extend arm forward
        current_joints[1] -= STEP
        current_joints[2] -= STEP
    elif "back" in cmd or "backward" in cmd:
        # Retract arm toward body
        current_joints[1] += STEP
        current_joints[2] += STEP

    # --- Diagonal Directions ---
    elif "diagonal up left" in cmd or "up left" in cmd:
        current_joints[0] += STEP  # Left
        current_joints[1] += STEP  # Up
        current_joints[2] -= STEP
    elif "diagonal up right" in cmd or "up right" in cmd:
        current_joints[0] -= STEP  # Right
        current_joints[1] += STEP  # Up
        current_joints[2] -= STEP
    elif "diagonal down left" in cmd or "down left" in cmd:
        current_joints[0] += STEP  # Left
        current_joints[1] -= STEP  # Down
        current_joints[2] += STEP
    elif "diagonal down right" in cmd or "down right" in cmd:
        current_joints[0] -= STEP  # Right
        current_joints[1] -= STEP  # Down
        current_joints[2] += STEP

    # --- Gripper Actions ---
    elif "open" in cmd or "release" in cmd:
        current_joints[5] = 30
    elif "close" in cmd or "grab" in cmd:
        current_joints[5] = 140

    # --- Reset / Home ---
    elif "home" in cmd or "reset" in cmd or "center" in cmd:
        current_joints = [90, 90, 90, 90, 90, 30]

    else:
        print("Command not recognized. Try: up, down, left, right, forward, back, diagonal, grab, release, home.")
        return

    apply_pose(400)
    print(f"Current angles: {current_joints}")

# ----------------------------------------------------
# Main Audio Listener Loop
# ----------------------------------------------------
recognizer = sr.Recognizer()
# Adjust sensitivity to ambient room noise
recognizer.energy_threshold = 4000
recognizer.dynamic_energy_threshold = True

print("\n--- Voice Control Active ---")
print("Available Voice Commands:")
print("• Cardinal: 'Up', 'Down', 'Left', 'Right', 'Forward', 'Back'")
print("• Diagonals: 'Up Left', 'Up Right', 'Down Left', 'Down Right'")
print("• Gripper: 'Grab' / 'Close', 'Release' / 'Open'")
print("• Utilities: 'Reset' / 'Home'")
print("Say 'exit' or press Ctrl+C to terminate.\n")

try:
    with sr.Microphone() as source:
        # Calibrate microphone against background static
        print("Calibrating mic for background noise... (stay quiet for 1 sec)")
        recognizer.adjust_for_ambient_noise(source, duration=1.0)
        print("Microphone ready. Speak a direction:")

        while True:
            try:
                # Capture phrase
                audio = recognizer.listen(source, timeout=5, phrase_time_limit=3)
                # Transcribe using Google's speech recognition endpoint
                spoken_text = recognizer.recognize_google(audio)
                print(f"\nHear: '{spoken_text}'")

                if "exit" in spoken_text.lower() or "quit" in spoken_text.lower():
                    print("Exiting voice control...")
                    break

                execute_voice_command(spoken_text)

            except sr.WaitTimeoutError:
                # Normal timeout when no speech occurs within 5s
                pass
            except sr.UnknownValueError:
                # Ambient noise was heard but no intelligible words
                pass
            except sr.RequestError as e:
                print(f"Speech recognition service error: {e}")

except KeyboardInterrupt:
    print("\nProgram stopped by user.")
finally:
    # Safely release arm controller
    del Arm
```

```python
import cv2
import numpy as np
import ipywidgets as widgets
from IPython.display import display
from PIL import Image
import io

# Setup Camera (tries index 0, then 1, then CSI pipeline)
def open_cam():
    for idx in [0, 1]:
        c = cv2.VideoCapture(idx, cv2.CAP_V4L2)
        if c.isOpened():
            ret, _ = c.read()
            if ret:
                c.set(cv2.CAP_PROP_FRAME_WIDTH, 320)
                c.set(cv2.CAP_PROP_FRAME_HEIGHT, 240)
                return c
            c.release()
    return None

cap = open_cam()
if cap is None:
    raise RuntimeError("Camera not detected or locked by another kernel.")

# Sliders for HSV Lower and Upper bounds
h_min = widgets.IntSlider(value=0, min=0, max=180, description='H Min:')
s_min = widgets.IntSlider(value=100, min=0, max=255, description='S Min:')
v_min = widgets.IntSlider(value=100, min=0, max=255, description='V Min:')

h_max = widgets.IntSlider(value=10, min=0, max=180, description='H Max:')
s_max = widgets.IntSlider(value=255, min=0, max=255, description='S Max:')
v_max = widgets.IntSlider(value=255, min=0, max=255, description='V Max:')

run_toggle = widgets.ToggleButton(value=True, description='Stop Camera', button_style='danger')
image_widget = widgets.Image(format='jpeg', width=320, height=240)
mask_widget = widgets.Image(format='jpeg', width=320, height=240)

# Layout controls
controls = widgets.VBox([
    widgets.HBox([h_min, h_max]),
    widgets.HBox([s_min, s_max]),
    widgets.HBox([v_min, v_max]),
    run_toggle
])
display(widgets.HBox([image_widget, mask_widget]), controls)

try:
    while run_toggle.value:
        ret, frame = cap.read()
        if not ret:
            break

        hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
        lower = np.array([h_min.value, s_min.value, v_min.value])
        upper = np.array([h_max.value, s_max.value, v_max.value])
        
        mask = cv2.inRange(hsv, lower, upper)
        result = cv2.bitwise_and(frame, frame, mask=mask)

        # Convert OpenCV BGR to JPEG for Jupyter display
        _, buf_frame = cv2.imencode('.jpg', frame)
        _, buf_mask = cv2.imencode('.jpg', result)
        
        image_widget.value = buf_frame.tobytes()
        mask_widget.value = buf_mask.tobytes()

finally:
    cap.release()
    print(f"\nFinal HSV Bounds:")
    print(f"Lower: [{h_min.value}, {s_min.value}, {v_min.value}]")
    print(f"Upper: [{h_max.value}, {s_max.value}, {v_max.value}]")
```

```python
import cv2
import numpy as np
import time
import ipywidgets as widgets
from IPython.display import display
from Arm_Lib import Arm_Device

arm = Arm_Device()
time.sleep(0.5)

POS_REST    = [90, 90, 90, 90, 90, 30]
POS_INSPECT = [90, 80, 50, 40, 90, 30]
POS_PREPICK = [90, 65, 40, 40, 90, 30]
POS_PICK    = [90, 53, 33, 36, 90, 155]

POS_DROP = {
    'red':    [35,  65, 45, 50, 90, 30],
    'green':  [65,  65, 45, 50, 90, 30],
    'blue':   [115, 65, 45, 50, 90, 30],
    'yellow': [145, 65, 45, 50, 90, 30]
}

# Wider HSV tolerances to handle room lighting variations
COLOR_RANGES = {
    'red': [
        (np.array([0, 100, 50]), np.array([10, 255, 255])),
        (np.array([165, 100, 50]), np.array([180, 255, 255]))
    ],
    'green': [
        (np.array([35, 60, 50]), np.array([85, 255, 255]))
    ],
    'blue': [
        (np.array([95, 70, 50]), np.array([135, 255, 255]))
    ],
    'yellow': [
        (np.array([18, 70, 70]), np.array([35, 255, 255]))
    ]
}

def move_servos(joints, duration_ms):
    arm.Arm_serial_servo_write6(
        int(joints[0]), int(joints[1]), int(joints[2]),
        int(joints[3]), int(joints[4]), int(joints[5]),
        int(duration_ms)
    )
    time.sleep((duration_ms / 1000.0) + 0.05)

def pick_and_place(color):
    if color not in POS_DROP:
        return
    move_servos(POS_PREPICK, 1000)
    arm.Arm_serial_servo_write6(POS_PICK[0], POS_PICK[1], POS_PICK[2], POS_PICK[3], POS_PICK[4], 30, 700)
    time.sleep(0.75)
    arm.Arm_serial_servo_write(6, 155, 500)
    time.sleep(0.6)
    arm.Arm_serial_servo_write6(POS_PREPICK[0], POS_PREPICK[1], POS_PREPICK[2], POS_PREPICK[3], POS_PREPICK[4], 155, 700)
    time.sleep(0.75)
    target = POS_DROP[color]
    arm.Arm_serial_servo_write6(target[0], target[1], target[2], target[3], target[4], 155, 1200)
    time.sleep(1.25)
    arm.Arm_serial_servo_write(6, 30, 400)
    time.sleep(0.5)
    move_servos(POS_INSPECT, 1000)

def detect_color(frame):
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    best_color = None
    max_area = 0
    best_box = None

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

    for color_name, ranges in COLOR_RANGES.items():
        mask = np.zeros(hsv.shape[:2], dtype=np.uint8)
        for lower, upper in ranges:
            mask = cv2.bitwise_or(mask, cv2.inRange(hsv, lower, upper))

        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

        contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        for cnt in contours:
            area = cv2.contourArea(cnt)
            # Reduced area requirement to 800 to detect smaller/further blocks
            if area > 800 and area > max_area:
                max_area = area
                best_color = color_name
                best_box = cv2.boundingRect(cnt)

    return best_color, best_box, max_area

# Force open index 0 with V4L2 directly
cap = cv2.VideoCapture(0, cv2.CAP_V4L2)
if not cap.isOpened():
    cap = cv2.VideoCapture(1, cv2.CAP_V4L2)

img_widget = widgets.Image(format='jpeg', width=320, height=240)
status_label = widgets.Label(value="Running diagnostics...")
stop_button = widgets.ToggleButton(value=False, description='Stop')

display(widgets.VBox([img_widget, status_label, stop_button]))
move_servos(POS_INSPECT, 1200)

consecutive_hits = 0
current_color = None

try:
    while not stop_button.value:
        ret, frame = cap.read()
        if not ret:
            status_label.value = "Camera frame drop."
            time.sleep(0.1)
            continue

        color, box, area = detect_color(frame)

        if box is not None:
            x, y, w, h = box
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
            cv2.putText(frame, f"{color.upper()} (Area: {int(area)})", (x, max(20, y - 8)),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

            if color == current_color:
                consecutive_hits += 1
            else:
                current_color = color
                consecutive_hits = 0

            status_label.value = f"Detected: {color} | Frames: {consecutive_hits}/8 | Area: {int(area)}"

            # Trigger on 8 consecutive frames
            if consecutive_hits >= 8:
                status_label.value = f"Executing pick for {color}..."
                _, buf = cv2.imencode('.jpg', frame)
                img_widget.value = buf.tobytes()
                pick_and_place(color)
                consecutive_hits = 0
                current_color = None
        else:
            consecutive_hits = 0
            current_color = None
            status_label.value = "No colored object detected in view."

        _, buf = cv2.imencode('.jpg', frame)
        img_widget.value = buf.tobytes()
        time.sleep(0.02)

finally:
    move_servos(POS_REST, 1000)
    cap.release()
    status_label.value = "Stopped."
```
