```python
# Voice Setup
model = Model("vosk-model-small-en-us-0.15")

q = queue.Queue()

def callback(indata, frames, time, status):
    q.put(bytes(indata))

mic_index = 3
device_info = sd.query_devices(mic_index, 'input')
samplerate = int(device_info['default_samplerate'])

rec = KaldiRecognizer(
    model,
    samplerate,
    '["home", "open","slope","back", "wave", "close", "dance","moon", "left right", "reset", "exit" ]'
)
```

```python
running = True

last_command = ""

def execute_command(text):

    global running
    global last_command

    text = text.lower().strip()

    if text == last_command:
        return

    last_command = text

    print("Command :", text)

    if "home" in text:
        home()

    elif "open" in text:
        open_gripper()

    elif "close" in text:
        close_gripper()

    elif "wave" in text:
        wave()
        
    elif "slope" in text:
        slope()
    elif "moon" in text:
        moon()
        
    elif "back" in text:
        back()

    elif "dance" in text:
        dance()

    elif "left right" in text:
        left_right()

    elif "stop" in text:
        home()

    elif "exit" in text:
        print("Stopping...")
        home()
        running = False

    else:
        print("Unknown command")
```

```python
with sd.RawInputStream(
    device=mic_index,
    samplerate=samplerate
    blocksize=4096
    dtype='int16',
    channels=1,
    callback=callback):
    print("----------------------------------")
    print("Voice Control Started")
    print("----------------------------------")
    print("Commands:")
    print("home")
    print("open")
    print("close")
    print("wave")
    print("dance")
    print("left right")
    print("stop")
    print("slope")
    print("back")
    print("moon")
    print("exit")
    print("----------------------------------")

    while running:
        data = q.get()

        if rec.AcceptWaveform(data):
            result = json.loads(rec.Result())
            text = result.get("text", "")

            if text:
                print("🧠 Heard:", text)
                execute_command(text)

print("Listener stopped")
home()

del Arm
```

```python
#!/usr/bin/env python3
# -*- coding:utf-8 -*-

import os
import json
import time
import queue
import sounddevice as sd
from vosk import Model, KaldiRecognizer
import Arm_Lib

# ---------------------------------------------------------
# 1. Arm Setup & Movement Definitions
# ---------------------------------------------------------
Arm = Arm_Lib.Arm_Device()
HOME = [90, 135, 0, 0, 90, 10]

Arm.Arm_serial_servo_write6_array(HOME, 1000)
time.sleep(1)

def home():
    print("Action: Home")
    Arm.Arm_serial_servo_write6_array(HOME, 1000)
    time.sleep(1.0)

def open_gripper():
    print("Action: Open Gripper")
    Arm.Arm_serial_servo_write(6, 60, 500)
    time.sleep(0.6)

def close_gripper():
    print("Action: Close Gripper")
    Arm.Arm_serial_servo_write(6, 180, 500)
    time.sleep(0.6)

def slope():
    print("Action: Slope")
    a = [90, 45, 90, 90, 90, 90]
    Arm.Arm_serial_servo_write6_array(a, 500)
    time.sleep(0.6)

def back():
    print("Action: Back")
    b = [90, 180, 0, 0, 90, 10]
    Arm.Arm_serial_servo_write6_array(b, 500)
    time.sleep(0.6)
    c = [90, 90, 90, 45, 45, 45]
    Arm.Arm_serial_servo_write6_array(c, 500)
    time.sleep(0.6)

def moon():
    print("Action: Moon")
    x = [31, 62, 34, 11, 30, 151]
    Arm.Arm_serial_servo_write6_array(x, 1000)
    time.sleep(1)
    Arm.Arm_serial_servo_write6(37, 62, 34, 11, 40, 70, 1000)
    time.sleep(1)
    Arm.Arm_serial_servo_write6(50, 62, 34, 11, 60, 151, 1000)
    time.sleep(1)
    y = [86, 62, 29, 11, 90, 70]
    Arm.Arm_serial_servo_write6_array(y, 1000)
    time.sleep(1)
    Arm.Arm_serial_servo_write6(95, 62, 34, 11, 100, 151, 1000)
    time.sleep(1)
    Arm.Arm_serial_servo_write6(106, 62, 34, 11, 110, 71, 1000)
    time.sleep(1)
    z = [121, 62, 34, 11, 120, 151]
    Arm.Arm_serial_servo_write6_array(z, 1000)
    time.sleep(1)
    slope()
    time.sleep(0.8)
    back()

def wave():
    print("Action: Wave")
    pose1 = [90, 90, 45, 45, 90, 10]
    pose2 = [90, 90, 45, 45, 130, 10]
    pose3 = [90, 90, 45, 45, 50, 10]

    Arm.Arm_serial_servo_write6_array(pose1, 800)
    time.sleep(0.9)

    for _ in range(3):
        Arm.Arm_serial_servo_write6_array(pose2, 500)
        time.sleep(0.6)
        Arm.Arm_serial_servo_write6_array(pose3, 500)
        time.sleep(0.6)

    home()

def dance():
    print("Action: Dance")
    Arm.Arm_serial_servo_write6(90, 90, 90, 90, 90, 90, 500)
    time.sleep(0.6)

    # Upper movements
    Arm.Arm_serial_servo_write(2, 60, 500)
    Arm.Arm_serial_servo_write(3, 120, 500)
    Arm.Arm_serial_servo_write(4, 60, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 45, 500)
    Arm.Arm_serial_servo_write(3, 135, 500)
    Arm.Arm_serial_servo_write(4, 45, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 60, 500)
    Arm.Arm_serial_servo_write(3, 120, 500)
    Arm.Arm_serial_servo_write(4, 60, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 90, 500)
    Arm.Arm_serial_servo_write(3, 90, 500)
    Arm.Arm_serial_servo_write(4, 90, 500)
    time.sleep(0.6)

    # Lower movements
    Arm.Arm_serial_servo_write(2, 100, 500)
    Arm.Arm_serial_servo_write(3, 80, 500)
    Arm.Arm_serial_servo_write(4, 80, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 120, 500)
    Arm.Arm_serial_servo_write(3, 60, 500)
    Arm.Arm_serial_servo_write(4, 60, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 135, 500)
    Arm.Arm_serial_servo_write(3, 45, 500)
    Arm.Arm_serial_servo_write(4, 45, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(2, 90, 500)
    Arm.Arm_serial_servo_write(3, 90, 500)
    Arm.Arm_serial_servo_write(4, 90, 500)
    time.sleep(0.6)

    # Gripper & wrist
    for _ in range(2):
        Arm.Arm_serial_servo_write(4, 20, 500)
        Arm.Arm_serial_servo_write(6, 150, 500)
        time.sleep(0.6)
        Arm.Arm_serial_servo_write(4, 90, 500)
        Arm.Arm_serial_servo_write(6, 90, 500)
        time.sleep(0.6)

    # Base rotation & spin
    Arm.Arm_serial_servo_write(1, 0, 500)
    Arm.Arm_serial_servo_write(5, 0, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(3, 180, 500)
    Arm.Arm_serial_servo_write(4, 0, 500)
    time.sleep(0.6)

    Arm.Arm_serial_servo_write(6, 180, 500)
    time.sleep(0.6)
    Arm.Arm_serial_servo_write(6, 0, 1000)
    time.sleep(1.0)

    home()

def left_right():
    print("Action: Left Right")
    for _ in range(2):
        Arm.Arm_serial_servo_write(1, 50, 400)
        time.sleep(0.5)
        Arm.Arm_serial_servo_write(1, 130, 400)
        time.sleep(0.5)
    home()

# ---------------------------------------------------------
# 2. Audio Hardware Resolver
# ---------------------------------------------------------
def get_input_device():
    """Finds an operational audio input device index and sample rate."""
    devices = sd.query_devices()
    print("Scanning available audio devices:")
    
    selected_idx = None
    rate = 16000

    # Pass 1: Look for explicit input channels
    for idx, d in enumerate(devices):
        max_in = d.get('max_input_channels', 0)
        name = d.get('name', 'Unknown')
        print(f"  [{idx}] {name} (Inputs: {max_in}, Default SR: {d.get('default_samplerate')})")
        if max_in > 0 and selected_idx is None:
            # Prefer non-loopback devices if USB is present
            if "usb" in name.lower() or "mic" in name.lower():
                selected_idx = idx
                rate = int(d.get('default_samplerate', 16000))

    # Pass 2: If no USB/Mic named device prioritized, take any device with input channels
    if selected_idx is None:
        for idx, d in enumerate(devices):
            if d.get('max_input_channels', 0) > 0:
                selected_idx = idx
                rate = int(d.get('default_samplerate', 16000))
                break

    # Pass 3: Fall back to system default ALSA device (index None triggers PortAudio default)
    if selected_idx is None:
        print("[WARN] No explicit input channel flag detected. Falling back to default system input.")
        try:
            default_info = sd.query_devices(kind='input')
            rate = int(default_info.get('default_samplerate', 16000))
            return None, rate
        except Exception:
            raise RuntimeError(
                "No microphone detected! Please check:\n"
                "1. Is your USB microphone or camera firmly connected?\n"
                "2. Run 'arecord -l' in terminal to ensure the audio hardware is detected by Linux."
            )

    return selected_idx, rate

mic_index, samplerate = get_input_device()
print(f"\nTarget Input Device: {mic_index} | Sample Rate: {samplerate} Hz")

# ---------------------------------------------------------
# 3. Vosk Speech Model
# ---------------------------------------------------------
model_path = "vosk-model-small-en-us-0.15"
if not os.path.exists(model_path):
    raise FileNotFoundError(f"Model path '{model_path}' not found. Download or extract the Vosk model in this folder.")

model = Model(model_path)
grammar = '["home", "open", "slope", "back", "wave", "close", "dance", "moon", "left right", "stop", "exit"]'
rec = KaldiRecognizer(model, samplerate, grammar)

audio_queue = queue.Queue()

def audio_callback(indata, frames, time_info, status):
    audio_queue.put(bytes(indata))

def clear_audio_queue():
    while not audio_queue.empty():
        try:
            audio_queue.get_nowait()
        except queue.Empty:
            break

# ---------------------------------------------------------
# 4. Command Router
# ---------------------------------------------------------
running = True

def execute_command(text):
    global running
    text = text.lower().strip()
    if not text:
        return

    print(f">> Heard command: '{text}'")

    if "home" in text:
        home()
    elif "open" in text:
        open_gripper()
    elif "close" in text:
        close_gripper()
    elif "wave" in text:
        wave()
    elif "slope" in text:
        slope()
    elif "moon" in text:
        moon()
    elif "back" in text:
        back()
    elif "dance" in text:
        dance()
    elif "left right" in text:
        left_right()
    elif "stop" in text:
        home()
    elif "exit" in text:
        print("Stopping listener...")
        home()
        running = False

    clear_audio_queue()

# ---------------------------------------------------------
# 5. Execution Loop
# ---------------------------------------------------------
print("---------------------------------------")
print("DOFBOT Voice Controller Ready")
print("Available: home | open | close | wave | dance | left right | slope | back | moon | stop | exit")
print("---------------------------------------")

try:
    with sd.RawInputStream(
        device=mic_index,
        samplerate=samplerate,
        blocksize=4096,
        dtype='int16',
        channels=1,
        callback=audio_callback
    ):
        while running:
            try:
                data = audio_queue.get(timeout=1.0)
            except queue.Empty:
                continue

            if rec.AcceptWaveform(data):
                result = json.loads(rec.Result())
                cmd = result.get("text", "")
                if cmd:
                    execute_command(cmd)

except KeyboardInterrupt:
    print("\nExecution stopped by user.")

finally:
    print("Safely homing arm...")
    home()
    time.sleep(0.5)
    try:
        del Arm
    except Exception:
        pass
    print("Exited cleanly.")
```

```python
#!/usr/bin/env python3
# -*- coding:utf-8 -*-

import os
import json
import time
import queue
import sounddevice as sd
from vosk import Model, KaldiRecognizer
import Arm_Lib

# -------------------------------------------------------------
# 1. Hardware Initialization
# -------------------------------------------------------------
arm = Arm_Lib.Arm_Device()
time.sleep(0.3)

# Core Servo Poses [S1 (Base), S2, S3, S4, S5 (Wrist Roll), S6 (Gripper)]
POSE_HOME   = [90, 135, 0, 10, 90, 30]
POSE_REST   = [90, 90, 90, 90, 90, 30]
POSE_STAND  = [90, 90, 0, 0, 90, 30]
POSE_CROUCH = [90, 45, 120, 15, 90, 30]

# Tracking current joint positions
current_joints = list(POSE_HOME)

def move_all(s1, s2, s3, s4, s5, s6, duration=800):
    """Executes joint array update and synchronizes internal state."""
    global current_joints
    angles = [
        max(0, min(180, s1)),
        max(0, min(180, s2)),
        max(0, min(180, s3)),
        max(0, min(180, s4)),
        max(0, min(180, s5)),
        max(0, min(180, s6))
    ]
    arm.Arm_serial_servo_write6_array(angles, duration)
    current_joints = angles
    time.sleep((duration / 1000.0) + 0.1)

def move_single(servo_id, target_angle, duration=500):
    """Updates a single joint while preserving the others."""
    global current_joints
    angle = max(0, min(180, target_angle))
    arm.Arm_serial_servo_write(servo_id, angle, duration)
    current_joints[servo_id - 1] = angle
    time.sleep((duration / 1000.0) + 0.05)

# Move to home on launch
move_all(*POSE_HOME, duration=1000)

# -------------------------------------------------------------
# 2. Movement Functions
# -------------------------------------------------------------
def action_home():
    print("[Action] Home Pose")
    move_all(*POSE_HOME, 1000)

def action_turn_left():
    print("[Action] Turning Base Left")
    move_single(1, 150, 700)

def action_turn_right():
    print("[Action] Turning Base Right")
    move_single(1, 30, 700)

def action_center():
    print("[Action] Centering Base")
    move_single(1, 90, 600)

def action_move_up():
    print("[Action] Elevating Arm")
    move_all(current_joints[0], 115, 20, 30, current_joints[4], current_joints[5], 800)

def action_move_down():
    print("[Action] Lowering Arm")
    move_all(current_joints[0], 55, 60, 45, current_joints[4], current_joints[5], 800)

def action_reach_forward():
    print("[Action] Reaching Forward")
    move_all(current_joints[0], 65, 30, 30, 90, current_joints[5], 900)

def action_pull_back():
    print("[Action] Pulling Back")
    move_all(current_joints[0], 130, 40, 20, 90, current_joints[5], 900)

def action_stand():
    print("[Action] Standing Tall")
    move_all(*POSE_STAND, 1000)

def action_crouch():
    print("[Action] Crouching Down")
    move_all(*POSE_CROUCH, 1000)

def action_look_up():
    print("[Action] Tilting Wrist Up")
    move_single(4, 0, 500)

def action_look_down():
    print("[Action] Tilting Wrist Down")
    move_single(4, 110, 500)

def action_slope():
    print("[Action] Sloping")
    move_all(current_joints[0], 45, 90, 90, 90, current_joints[5], 800)

def action_twist_left():
    print("[Action] Rolling Wrist Counter-Clockwise")
    move_single(5, 0, 500)

def action_twist_right():
    print("[Action] Rolling Wrist Clockwise")
    move_single(5, 180, 500)

def action_twist_flat():
    print("[Action] Leveling Wrist Roll")
    move_single(5, 90, 500)

def action_open_gripper():
    print("[Action] Opening Gripper")
    move_single(6, 30, 400)

def action_close_gripper():
    print("[Action] Closing Gripper")
    move_single(6, 165, 500)

def action_half_open():
    print("[Action] Setting Gripper Half Open")
    move_single(6, 80, 400)

def action_nod():
    print("[Action] Nodding")
    for _ in range(2):
        move_single(4, 30, 300)
        move_single(4, 90, 300)

def action_wave():
    print("[Action] Waving")
    move_all(90, 90, 45, 45, 90, 30, 800)
    for _ in range(3):
        move_single(5, 140, 350)
        move_single(5, 40, 350)
    action_home()

def action_bow():
    print("[Action] Bowing")
    move_all(current_joints[0], 45, 45, 20, 90, current_joints[5], 1000)
    time.sleep(0.5)
    action_home()

def action_dance():
    print("[Action] Dancing")
    move_all(90, 90, 90, 90, 90, 90, 600)
    move_single(2, 60, 400)
    move_single(3, 120, 400)
    move_single(1, 45, 500)
    move_single(1, 135, 700)
    move_single(1, 90, 500)
    for _ in range(2):
        move_single(6, 150, 300)
        move_single(6, 30, 300)
    action_home()

def action_sleep():
    print("[Action] Relaxing Motors (Torque Off)")
    arm.Arm_serial_set_torque(0)

# -------------------------------------------------------------
# 3. Audio & Vosk Configuration
# -------------------------------------------------------------
def resolve_mic_device():
    """Detects available microphone hardware without crashing."""
    devices = sd.query_devices()
    selected_idx = None
    rate = 16000

    # Search for an audio input device
    for idx, d in enumerate(devices):
        if d.get('max_input_channels', 0) > 0:
            name = d.get('name', '').lower()
            if "usb" in name or "mic" in name or "camera" in name:
                return idx, 16000
            if selected_idx is None:
                selected_idx = idx

    if selected_idx is not None:
        return selected_idx, 16000

    # If PortAudio returned zero query channels, attempt default input
    try:
        def_in = sd.query_devices(kind='input')
        return None, int(def_in.get('default_samplerate', 16000))
    except Exception:
        raise RuntimeError(
            "No audio input device detected by system.\n"
            "Please plug a USB microphone or webcam with mic into the DOFBOT."
        )

mic_index, samplerate = resolve_mic_device()
print(f"[Audio] Selected Mic Device Index: {mic_index} | Rate: {samplerate} Hz")

model_dir = "vosk-model-small-en-us-0.15"
if not os.path.exists(model_dir):
    raise FileNotFoundError(f"Vosk model directory '{model_dir}' not found in current folder.")

vosk_model = Model(model_dir)

# Grammar dictionary matching recognized commands
COMMAND_GRAMMAR = [
    "home", "turn left", "turn right", "center",
    "reach forward", "pull back", "move up", "move down",
    "stand tall", "crouch", "look up", "look down",
    "slope forward", "twist left", "twist right", "twist flat",
    "open gripper", "close gripper", "half open",
    "wave", "nod", "bow", "dance", "sleep", "exit"
]

rec = KaldiRecognizer(vosk_model, samplerate, json.dumps(COMMAND_GRAMMAR))
audio_queue = queue.Queue()

def audio_stream_callback(indata, frames, time_info, status):
    audio_queue.put(bytes(indata))

def flush_audio_queue():
    """Clears accumulated audio so servo noise doesn't trigger ghost commands."""
    while not audio_queue.empty():
        try:
            audio_queue.get_nowait()
        except queue.Empty:
            break

# -------------------------------------------------------------
# 4. Command Dispatcher
# -------------------------------------------------------------
running = True

def parse_and_execute(text):
    global running
    text = text.lower().strip()
    if not text:
        return

    print(f"\n[Voice Heard] >> '{text}'")

    if text == "home":
        action_home()
    elif text == "turn left":
        action_turn_left()
    elif text == "turn right":
        action_turn_right()
    elif text == "center":
        action_center()
    elif text == "reach forward":
        action_reach_forward()
    elif text == "pull back":
        action_pull_back()
    elif text == "move up":
        action_move_up()
    elif text == "move down":
        action_move_down()
    elif text == "stand tall":
        action_stand()
    elif text == "crouch":
        action_crouch()
    elif text == "look up":
        action_look_up()
    elif text == "look down":
        action_look_down()
    elif text == "slope forward":
        action_slope()
    elif text == "twist left":
        action_twist_left()
    elif text == "twist right":
        action_twist_right()
    elif text == "twist flat":
        action_twist_flat()
    elif text == "open gripper":
        action_open_gripper()
    elif text == "close gripper":
        action_close_gripper()
    elif text == "half open":
        action_half_open()
    elif text == "nod":
        action_nod()
    elif text == "wave":
        action_wave()
    elif text == "bow":
        action_bow()
    elif text == "dance":
        action_dance()
    elif text == "sleep":
        action_sleep()
    elif text == "exit":
        print("[System] Shutting down voice control...")
        action_home()
        running = False

    # Discard audio recorded during movement execution
    flush_audio_queue()

# -------------------------------------------------------------
# 5. Continuous Listening Loop
# -------------------------------------------------------------
print("\n" + "="*50)
print(" DOFBOT MULTI-DIRECTION VOICE SYSTEM ACTIVE")
print("="*50)
print("Speak clearly into your microphone.")
print("Say 'exit' anytime to stop the program.\n")

try:
    with sd.RawInputStream(
        device=mic_index,
        samplerate=samplerate,
        blocksize=4096,
        dtype='int16',
        channels=1,
        callback=audio_stream_callback
    ):
        while running:
            try:
                raw_chunk = audio_queue.get(timeout=1.0)
            except queue.Empty:
                continue

            if rec.AcceptWaveform(raw_chunk):
                data = json.loads(rec.Result())
                command = data.get("text", "")
                if command:
                    parse_and_execute(command)

except KeyboardInterrupt:
    print("\n[System] Keyboard interruption detected.")

finally:
    print("[System] Restoring arm to home and releasing interface...")
    action_home()
    time.sleep(0.5)
    try:
        del arm
    except Exception:
        pass
    print("[System] Done.")
```
